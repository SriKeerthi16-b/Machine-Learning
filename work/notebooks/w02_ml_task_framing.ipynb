

### Task Type: Ranking / Scoring

The goal is to identify webpages that may need content refresh or further review.

Instead of applying a fixed rule to every page, the ML system assigns each page a score indicating how likely it is to be a content-refresh opportunity.

The pages can then be ranked from highest priority to lowest priority.

### ML Decision

The practical decision is:

> Which pages should a content team review first?

The output is therefore a ranked list of webpages for human review.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)
```

```python
df["is_declining_label"].value_counts()
```



The target is whether a page is showing a declining trend.

I define the target as:

`is_declining_label = trend_direction == "down"`

This target acts as a proxy for identifying pages that may require content-refresh attention.

This does not mean that every declining page definitely needs a refresh. The model is used to prioritize pages for human review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Baseline: overall proportion of pages labeled as declining
baseline_precision = df["is_declining_label"].mean()

print(f"Declining-page base rate: {baseline_precision:.3f}")
print(f"Baseline Precision@50: {baseline_precision:.3f}")
```



The main success metric is Precision@K, with K = 50.

Precision@50 answers:

> Of the 50 pages ranked highest by the model, what fraction are actually labeled as declining?

The baseline is the overall proportion of declining pages in the dataset. A useful scoring system should achieve a Precision@50 higher than this baseline.

This metric fits the task because the content team has limited review capacity and needs the most relevant pages near the top of the list.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*



A fixed rule could say that a page should be reviewed when one metric crosses a threshold, such as impressions falling below a certain value.

However, webpage performance depends on multiple signals.

Different pages can have different combinations of impressions, clicks, CTR, position, content age, and other measurements.
