### Search question

Which pages should be reviewed first for a potential content refresh or another appropriate action, based on observable search and engagement signals?

### Unit of analysis

The unit of analysis is a **page**. Each row represents a page and contains observable signals about its search performance and content characteristics.

### Decision

The decision is **which pages should receive human review first**.

### Action

A reviewer could use the ranking to decide whether a page should be refreshed, expanded, investigated for a possible CTR issue, protected, monitored, or considered for pruning or consolidation when supported by evidence.

The model would support the decision rather than automatically making the content change.

### Cost of a wrong recommendation

A false positive could cause a team to spend limited time reviewing a page that does not need attention.

A false negative could cause a page that needs attention to be missed or reviewed later than it should be.

Therefore, the system should be treated as decision support rather than an automatic content-changing system.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import subprocess
import pandas as pd
import numpy as np

# Go to /content
os.chdir("/content")

# Clone the starter repository if needed
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if not os.path.isdir(REPO_DIR):
    subprocess.run(
        ["git", "clone", "--depth", "1", REPO_URL, REPO_DIR],
        check=True
    )

# Go into the repository
os.chdir(REPO_DIR)

# Load the dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create the declining label
df["is_declining_label"] = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .astype(int)
)

# Create the hand-rule score
stale = (df["days_since_last_update"] >= 180).astype(int)
visible = (df["impressions_90d"] >= 500).astype(int)

df["hand_rule_score"] = (
    stale * visible * df["impressions_90d"]
)

# Precision@K function
def precision_at_k(scores, labels, k):
    order = np.argsort(-np.asarray(scores))
    topk = np.asarray(labels)[order[:k]]
    return topk.mean()

# Calculate Precision@50
precision_50 = precision_at_k(
    df["hand_rule_score"],
    df["is_declining_label"],
    50
)

# Display results
print("Number of pages:", len(df))
print("Number of columns:", df.shape[1])
print("Declining rate:", round(df["is_declining_label"].mean(), 3))
print("Hand rule Precision@50:", round(precision_50, 3))
```
